```python

import matplotlib.pyplot as plt
import numpy as np

# X ko 2D format mein rakho
X = np.array(X).reshape(-1, 1)

# Train Linear Regression model
lr.fit(X, y)

# Blue dots: Actual data
plt.scatter(X, y, color="blue", label="Actual Data")

# Green line: Predicted values
x_line = np.sort(X, axis=0)
y_pred = lr.predict(x_line)

plt.plot(x_line, y_pred, color="green", label="Regression Line")

# Graph details
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Graph")
plt.legend()
plt.grid(True)

plt.show()
```
